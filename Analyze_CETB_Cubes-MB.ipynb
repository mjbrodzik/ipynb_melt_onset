### Brightness Temperature (Tb) Threshold Optimization Algorithm for Snowmelt Detection

### Key Features:
1. Handles Different Snow Classes:  
   - Optimizes thresholds based on the local climate and snowmelt dynamics.  
   - Snow classes are processed differently to focus on their distinct melt periods.
   - For Snow Classes: ['Boreal Forest', 'Montane Forest', 'Tundra', 'Prairie'] >> Time Frame: January to September >>> For regions where snow is a defining feature of the landscape and climate, capturing the entirety of a snow and melt season within a single year is crucial. Empirical trials have shown that focusing on the January to September period allows for more insightful analysis of the snow and melt cycle. By excluding the months of October to December, the data analysis sharpens its focus on the active snowmelt period, reducing noise from the accumulation phase, which predominates in those months. Focusing on the critical snowmelt months from January to September allows for a better understanding of data variations due to environmental factors like temperature fluctuations and solar radiation, which significantly influence the melting processes. This period covers winter, spring, and part of summer, including the tail end of the Arctic winter, the entire spring thaw, and most of the summer melt season. As a result, the histograms generated during this time will primarily reflect the brightness temperatures associated with these conditions, showcasing a range of values from very cold (frozen conditions) to warmer (surface melting). By excluding October to December, the histograms will not capture the re-freezing period and the onset of the new snow season. This helps in emphasizing the melting phase, aiding in more precise identification of the onset of snowmelt.
   - For Snow Classes: ['Maritime', 'Ephemeral', 'Ice', 'Ocean'] >> Time Frame: Calendar Year (January to December) >>> For regions where snow is not the primary climatic feature or where it is less predictable, using the full calendar year (January to December) provides a consistent and straightforward timeframe for analysis. In these areas, where transient or less defining snow events occur, capturing long-term trends can be more beneficial than focusing solely on specific snowmelt periods.
   - Characteristics of Different Snow Classes:
     1) Tundra: Long, harsh winters and short summers, requiring a focused analysis on the snowmelt period.
     2) Boreal Forest: Dominated by long, cold winters, where snow remains a significant factor for much of the year.
     3) Montane Forest: Diverse snow conditions due to varied elevations and aspects, making it important to study the snow season in its entirety.
     4) Prairie: Experiences unpredictable snow events; capturing the entire snow cycle from the first snowfall to complete melt is crucial.
     5) Maritime: Milder climates with more transient snow events, benefiting from a broader analysis over the calendar year.
     6) Ephemeral: Sporadic snowfall that does not last long, making it suitable to consider the entire year.
     7) Ice: Persistent ice cover throughout the year, with long-term trends often more relevant than seasonal changes.
     8) Ocean: No defined snow season, where ice and snow accumulation are best analyzed over the calendar year due to their long-term implications for climate change.
    - Snow classes are derived from Global Seasonal-Snow Classification, Version >>> Liston, G. E. & Sturm, M. (2021). Global Seasonal-Snow Classification. (NSIDC-0768, Version 1). [Data Set]. Boulder, Colorado USA. NASA National Snow and Ice Data Center Distributed Active Archive Center. https://doi.org/10.5067/99FTCYYYLAQ0. [describe subset used if applicable]. Date Accessed 02-27-2025.

2. Applies Data Filtering & Processing Steps:  
   - Uses specific time frames to exclude irrelevant data and focus on snowmelt phases.  
   - Performs histogram smoothing and peak detection to identify key temperature transitions.  

3. Enhances Snowmelt Detection Accuracy:

### Methodology & Workflow for Tb Threshold Optimization Algorithm
#### Step 1: Extract Brightness Temperature Data (Based on Snow Class)
- The algorithm reads satellite brightness temperature (Tb) data and selects the relevant time period 
  based on the snow classification of the region.
- For Boreal Forest, Montane Forest, Tundra, and Prairie:  
  - Only data from January to September is used to avoid noise from the snow accumulation period.  
  - This ensures the histogram primarily reflects melting phases.  
- For Maritime, Ephemeral, Ice, and Ocean:  
  - The full calendar year (January to December) is used.  
  - This is because snowmelt in these regions is less predictable or persists year-round.

#### Step 2: Compute a Smoothed Histogram of Brightness Temperature
- A histogram is generated from the filtered Tb data.  
- Smoothing is applied using a Gaussian kernel to highlight significant temperature transitions.  
- The optimal kernel width is computed using Silverman’s rule, adjusted for this application.  
- This allows better identification of peaks and valleys in the temperature distribution.

#### Step 3: Detect Temperature Thresholds for Snowmelt
- Peak detection is performed to find key temperature transitions.  
- The 70th percentile of histogram values is used as a threshold to detect prominent peaks.  
- A distance parameter (determined via empirical testing) helps separate meaningful peaks.  


#### Step 4: Determine the Optimal Threshold Temperature
- If two peaks (bimodal distribution) are found, the minimum point between them is selected as the threshold.  
- If one peak (unimodal distribution) is found, a warning is issued, and manual review is needed.  
- If more than two peaks are found, an additional filtering step refines peak selection:  
  - Peaks are re-selected based on a mean-based threshold, defined as the mean of the histogram.  
  - Only peaks with a prominence exceeding the mean of the histogram are retained, eliminating minor fluctuations or noise.  
  - If more than three peaks remain after filtering, the histogram is flagged as complex and may require manual review.  
  - If exactly three prominent peaks remain, the optimal threshold is determined as the minimum valley point between the two most significant peaks.  


#### Step 5: Generate and Save Results
- A histogram plot is generated, displaying:  
  - The smoothed brightness temperature distribution  
  - The new optimized Tb threshold (dashed black line)  
  - The legacy 246K threshold (dashed red line) for comparison  
- The detected Tb threshold is saved for each site.

### DAV (Diurnal Amplitude Variations) Threshold Optimization Algorithm for Snowmelt Detection: Run the DAV algorithm for single-site optimization >>> The output shows the processing site, DAV Threshold (90th Percentile), DAV Threshold (95th Percentile), and DAV Threshold (Average of 90th and 95th percentile)

Author: Mahboubeh Boueshagh

This algorithm determines optimized DAV thresholds to detect the onset of snowmelt using CETB data. The 90th percentile of DAV is used as the threshold after validating that the DAV distribution is right-skewed, meaning most values are low, with occasional high spikes representing snowmelt events.

### Key Features:
1. Extracts DAV Data for a Specific Site and Year
   - Uses CETB data to calculate DAV.
   - Filters data for the year of interest and removes noise.

2. Calculates Percentile-Based Thresholds
   - Computes the 90th and 95th percentiles of DAV to capture extreme variations.
   - Uses the 90th percentile as the final DAV threshold, since testing showed it captures sudden increases in DAV and aligns with snowmelt onset while providing similar results to the 95th percentile without being overly restrictive. 

3. Visualizes DAV and Thresholds
   - Plots DAV time series with 90% and 95% thresholds as reference lines.
   - Saves the plot for further analysis and verification.


### Why Use the Percentile Method for Melt Detection?
- The DAV distribution is right-skewed, meaning most values are small, but there are occasional large spikes.
- These spikes represent sudden increases in DAV, which typically occur when snow begins to melt.
- If a normal distribution were used, the threshold would be influenced by the mean, which is affected by 
  the many low DAV values, leading to an unreliable threshold.
- By using the percentile method, we focus on the extreme values rather than the mean, which helps 
  isolate the melt signal from typical background variations.
- The 90th percentile was chosen because it effectively captures these spikes while avoiding false detections from smaller fluctuations. Additionally, testing showed that using the 90th percentile produces results similar to the 95th percentile without significantly altering melt detection.

### Note from tested techniques:
In the first try to optimize DAV algorithm that works best for a range of sites, I used the rate of change of DAV and its mean and std values to estimate the DAV threshold when it starts to increase during the melting season. I decided to do more analysis, and I made some adjustments to the algorithm. I checked the distribution of DAV and its rate of change. The DAV has a right-skewed distribution and its rate of change has a relatively normal distribution (look at the below chunk of code). When the DAV distribution is right-skewed, it means that there are more frequent smaller values with occasional larger spikes. With a longer tail on the right side, the majority of the data points will cluster on the lower end (representing typical non-melt days), while occasional higher spikes would indicate unique events, potentially snowmelt events. Thus, I decided to use the percentile method for melt detection. I tested the 90th and 95th percentiles, and after trial and error, and validation, I chose the 90th percentile as the DAV threshold for each site. However, the choice of the threshold depends on how conservative we want to be for melt detection.

## Avoiding Redundant Processing of rSIR Pixels in Each GRD Pixel
In this implementation, in order to have maximum data for histogram analysis we use the Tb data for all 64 pixels, which means it will repeat the same process with the same data for each rSIR pixel or site. To avoid this, we ensure that the Tb threshold computation is performed only once for each GRD pixel, even though multiple rSIR pixels exist within it. Instead of repeating the threshold calculation for every individual rSIR pixel, we implement an efficient caching mechanism using the dictionary `processed_GRD_pixels`. Please note that we can also run the process using Tb data relevant to the single rSIR pixel, but we may have data gaps or not a very continuous histogram. By using this method, we increase efficiency, reduce processing time, and ensure consistency in assigning Tb thresholds across all rSIR pixels within the same GRD pixel.
 

### How We Avoid Redundant Computation:
1. Identify Unique GRD Pixels  
2. Check if the GRD Pixel is Already Processed  
3. Compute Tb Threshold Only Once per GRD Pixel  
   - If a new GRD pixel is encountered, the script processes only one rSIR pixel within it.  
4. Assign the Same Threshold to All rSIR Pixels in That GRD Pixel  
   - Once a threshold is computed for a GRD pixel, it is directly assigned to all its corresponding rSIR pixels without reprocessing.

### Why This Approach is Efficient?
- Minimizes redundant computation → Each GRD pixel is processed only once, significantly reducing computational cost.  
- Faster execution → Instead of processing all rSIR pixels individually, the loop efficiently assigns precomputed values.  
- Memory optimization → Storing previously computed values in a dictionary prevents unnecessary function calls and data storage.  

```python
csv_filename = "thresholds2003_02.csv"

# List to store the calculated minimum temperatures (Tb threshold) for each site
min_temperatures = []

def main():
    processed_GRD_pixels = {}  # Dictionary to store calculated thresholds for each GRD pixel

    # Iterate over each site in the combined dataframe (holding site information from previous chunk)
    for index, row in site_data_combined.iterrows():
        Site = row['Site']
        lat_start = row['lat_start']
        lon_start = row['lon_start']
        x = row['x']
        y = row['y']
        snow_class = row['SnowClasses']
        
        # Compute the corresponding GRD pixel coordinates
        rows_cols_GRD = coords(datadir_GRD, prefix_GRD, lat_start, lat_start, lon_start, lon_start)
        grd_pixel_key = tuple(rows_cols_GRD)  # Unique identifier for each GRD pixel

        # If this GRD pixel has already been processed, assign the stored Tb threshold to all its rSIR pixels
        if grd_pixel_key in processed_GRD_pixels:
            tb_threshold = processed_GRD_pixels[grd_pixel_key]
            print(f"Assigning existing Tb threshold for Site {Site} in GRD {grd_pixel_key}: {tb_threshold}")
        else:
            # Process only one rSIR pixel within this GRD pixel
            print(f"Processing site: {Site}, Snow Class: {snow_class} for GRD {grd_pixel_key}")

            # Calculate the envelope of coordinates for 64 SIR pixels
            rows_cols_env = tuple(np.array(rows_cols_GRD) * sir_2_grd_factor)

            # Read brightness temperature data for the calculated coordinates from the SIR dataset
            data_SIR = read_Tb_whole(datadir_SIR, prefix_SIR, subYears, 
                                     rows_cols_env[0], rows_cols_env[1], 
                                     rows_cols_env[2], rows_cols_env[3])

            # Year of interest
            year = [2003]

            # Extract relevant data from data_SIR
            cal_year = data_SIR['cal_year']
            cal_month = data_SIR['cal_month']

            # Filter the data based on criteria like snow class and year
            data = extract_relevant_data(data_SIR, year, cal_year, cal_month, snow_class, Site)

            # Calculate a smoothed histogram for the brightness temperature data
            hist, bin_edges, hist_smooth = compute_smoothed_histogram(data, snow_class, Site)

            # Analyze the histogram to derive results: Tb thresholds
            results = analyze_histogram(hist, bin_edges, hist_smooth, Site, year, sensor_SIR, channel_SIR, snow_class, ThresholdDir)

            # Store the calculated Tb threshold
            tb_threshold = results['threshold']
            processed_GRD_pixels[grd_pixel_key] = tb_threshold  # Save for reuse

        # Assign the Tb threshold value to all rSIR pixels located in this GRD pixel
        min_temperatures.append([Site, snow_class, lat_start, lon_start, x, y, tb_threshold])

    # Save the computed Tb thresholds to a CSV file
    csv_path = os.path.join(ThresholdDir, csv_filename)
    with open(csv_path, 'w', newline='') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerow(['Site', 'Snow Class', 'Latitude', 'Longitude', 'x', 'y', 'Tb Threshold'])
        writer.writerows(min_temperatures)

    print(f"CSV file saved with minimum temperature values: {csv_path}")

# Execute the main function
if __name__ == "__main__":
    main()
```

```python
### This block is the optimized DAV algorithm for the region and year of interest!

# Subset of years
# It's important to have at least two years in 'subYears' to avoid certain errors, including the year of interest and the year after, Cont'd...
# for example, for 2003, you should also include 2004. 
subYears = Years[8:11]

# Set the year of interest
year = 2003

# Dictionary to store calculated DAV thresholds for each GRD pixel
processed_GRD_pixels = {}

# Function to extract and plot Tb and DAV for the specified year
def TbYear(CETB_data, DAV, cal_date, cal_ysear, year):
    """
    Extracts the time-series of Tb and DAV for the given year.

    Parameters:
        CETB_data: The CETB data array (brightness temperature data).
        DAV: The DAV data array (Diurnal Amplitude Variations Melt Algorithm).
        cal_date: Array of datetime objects corresponding to the CETB data.
        cal_year: Array of years corresponding to the CETB data.
        year: The year of interest to filter data.

    Returns:
        DataFrame containing the DAV data for the specified year and site.
    """
    # Create a list of pixel indices for the y and x dimensions
    y_dims_list = list(range(len(CETB_data[0, :, 0])))  # y-dimension pixel indices
    x_dims_list = list(range(len(CETB_data[0, 0, :])))  # x-dimension pixel indices

    # Boolean mask to filter data for the year of interest
    thisYr = cal_year == year

    # Iterate over all y and x pixel indices to plot the time series (if needed)
    for i in y_dims_list:
        for j in x_dims_list:
            # Create DataFrames for DAV time series for the selected year
            DAV_df = pd.DataFrame(DAV[thisYr, i, j], index=cal_date[thisYr])

    # Return the final DataFrame containing DAV for the specified year
    return DAV_df

# Initialize list to store DAV thresholds for each site
DAV_Threshold_Percentile = []

# Loop through all sites
for index, row in Sites_df.iterrows():
    # Extract site and coordinates
    Site = row['Site']
    lat_start = row['lat_start']
    lon_start = row['lon_start']
    x = row['x']
    y = row['y']

    # Compute the corresponding GRD pixel coordinates
    rows_cols_GRD = coords(datadir_GRD, prefix_GRD, lat_start, lat_start, lon_start, lon_start)
    grd_pixel_key = tuple(rows_cols_GRD)  # Unique identifier for each GRD pixel

    # If this GRD pixel has already been processed, assign the stored DAV threshold
    if grd_pixel_key in processed_GRD_pixels:
        DAV_threshold_Percentile90, DAV_threshold_Percentile95, DAV_Average = processed_GRD_pixels[grd_pixel_key]
        print(f"Assigning existing DAV thresholds for Site {Site} in GRD {grd_pixel_key}: "
              f"90%: {DAV_threshold_Percentile90:.2f} K, 95%: {DAV_threshold_Percentile95:.2f} K")
    else:
        print(f"Processing site: {Site} for GRD {grd_pixel_key}")

        # Calculate the envelope of coordinates for 64 SIR pixels
        rows_cols_env = tuple(np.array(rows_cols_GRD) * sir_2_grd_factor)

        # Read brightness temperature dataset
        data_SIR = read_Tb_whole(datadir_SIR, prefix_SIR, subYears, 
                                 rows_cols_env[0], rows_cols_env[1], 
                                 rows_cols_env[2], rows_cols_env[3])

        # Step 1: Calculate DAV using an existing function
        DAV_SIR = calc_DAV(data_SIR['TB'])

        # Step 2: Extract the Tb and DAV data for the specified year
        DAV_df = TbYear(data_SIR['TB'], DAV_SIR, data_SIR['cal_date'], data_SIR['cal_year'], year)

        # Step 3: Rename the DataFrame column to 'DAV' for clarity
        DAV_df.columns = ['DAV']

        # Step 4: Calculate the 90th and 95th percentiles of DAV values, as well as their average
        DAV_threshold_Percentile90 = DAV_df['DAV'].quantile(0.90)  # 90% threshold
        DAV_threshold_Percentile95 = DAV_df['DAV'].quantile(0.95)  # 95% threshold
        DAV_Average = np.mean([DAV_threshold_Percentile90, DAV_threshold_Percentile95])

        print(f"DAV Threshold (90% Percentile) at {Site}, Year {year}: {DAV_threshold_Percentile90:.2f} K")
        print(f"DAV Threshold (95% Percentile) at {Site}, Year {year}: {DAV_threshold_Percentile95:.2f} K")
        print(f"DAV Threshold (Average) at {Site}, Year {year}: {DAV_Average:.2f} K")

        # Store computed DAV thresholds in the dictionary
        processed_GRD_pixels[grd_pixel_key] = (DAV_threshold_Percentile90, DAV_threshold_Percentile95, DAV_Average)

    # Append the site and threshold to DAV_threshold list
    DAV_Threshold_Percentile.append((Site, lat_start, lon_start, x, y, DAV_threshold_Percentile90, DAV_threshold_Percentile95, DAV_Average))

# After processing all sites, save thresholds to a single CSV file
csv_path = os.path.join(ThresholdDir, f"DAVthresholds_AllSites_{year}.csv")
with open(csv_path, 'w', newline='') as csvfile:
    writer = csv.writer(csvfile)
    writer.writerow(['Site', 'Latitude', 'Longitude', 'x', 'y', 'DAV_Threshold_90_Percentile', 'DAV_Threshold_95_Percentile', 'DAV Threshold (Average)'])
    writer.writerows(DAV_Threshold_Percentile)

print(f"CSV file saved with DAV threshold values for all sites: {csv_path}")
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
import scipy.io as sio


# build the string that is the index to the SIR cell closest to our site
# note that this is likely inside the SIR cell envelope
rows_cols = coords(datadir_SIR, prefix_SIR, lat_start, lat_start, lon_start, lon_start)
Site_nearest_row_col = "%s,%s" % (str(rows_cols[0]),str(rows_cols[2])) 

#Define the path to save the plot
save_path = os.path.join(ThresholdDir, f"{Site}_{year}_{sensor_SIR}TimeSeries_StdPlot.png")

year = 2003

# Ensure the index is in datetime format
DAV_df.index = pd.to_datetime(DAV_df.index)

# reate a date range starting from October of the previous year to the end of the year of interest
start_date = f"{year}-01-01"
end_date = f"{year}-12-31"

# Filter the DataFrame for the specified date range
DAV_df = DAV_df[start_date:end_date]


# load dataframe of Tb GRD for plotting Tb on specified day and observation
#Tb_std_dev_GRD = read_Tb_std_dev(datadir_GRD, prefix_GRD, subYears,
                                 #rows_cols_GRD[0],rows_cols_GRD[1],rows_cols_GRD[2],rows_cols_GRD[3])
# slice of dataframe described above
#Tb_sd_frame_GRD=pd.DataFrame(Tb_std_dev_GRD[:,0,0], index=data_GRD['cal_date'])
#Tb_sd_frame_GRD.columns = ['GRD SSD']

# Ensure the index is in datetime format
#Tb_sd_frame_GRD.index = pd.to_datetime(Tb_sd_frame_GRD.index)

# Filter the DataFrame for the specified date range
#Tb_sd_frame_GRD = Tb_sd_frame_GRD[start_date:end_date]


#need to run this cell for the max DAV to work
y_dims_list=list(range(len(data_SIR['TB'][0,:,0])))  # creates a list of the y-dimension pixel indices, used for plotting
x_dims_list=list(range(len(data_SIR['TB'][0,0,:])))	# creates a list of the x-dimension pixel indices, for plotting
y_s=list(range(rows_cols_env[0],rows_cols_env[1]))  # makes a list of the y(row) numbers so the dataframe of early melt events can be indexed
x_s=list(range(rows_cols_env[2],rows_cols_env[3]))  # makes a list of the x(col) numbers so the dataframe of early melt events can be indexed

# load dataframe of Tb SIR for plotting Tb on specified day and observation
Tb_frame=pd.DataFrame()
for i in y_dims_list:
    for j in x_dims_list:
        column=pd.DataFrame(data=data_SIR['TB'][:,i,j], columns=[str(y_s[i])+','+str(x_s[j])])
        Tb_frame=pd.concat([Tb_frame,column],axis=1)
Tb_frame.index=data_SIR['cal_date']
# compute std dev of the 64 SIR pixels
Tb_std_64=Tb_frame.std(axis=1)
Tb_std=pd.DataFrame(Tb_std_64)
Tb_std.columns = ['SSD']

# Ensure the index is in datetime format
Tb_std.index = pd.to_datetime(Tb_std.index)


# Filter the DataFrame for the specified date range
Tb_std = Tb_std[start_date:end_date]
# Ensure the index is in datetime format
Tb_std.index = pd.to_datetime(Tb_std.index)

# Fill or interpolate missing values
Tb_std['SSD'] = Tb_std['SSD'].interpolate(method='linear')

Tb_nearest=Tb_frame[Site_nearest_row_col]
Tb_nearest=Tb_nearest.to_frame()
Tb_nearest.columns=['TB']
Tb_nearest.index = pd.to_datetime(Tb_nearest.index)
Tb_nearest = Tb_nearest[start_date:end_date]


MeltDate = pd.to_datetime('2003-04-13')


######### 
# Calculate the 15th percentile of the SSD data
percentile_60 = Tb_std['SSD'].quantile(0.6)
percentile_99 = Tb_std['SSD'].quantile(0.993)

# Decompose the time series to extract the trend
Tb_std_Clean = Tb_std['SSD'].dropna()
Tb_std_Clean = pd.DataFrame(Tb_std_Clean, columns=['SSD'])
decomposition = seasonal_decompose(Tb_std_Clean['SSD'], model='additive', period=30)  # Assuming daily data and monthly seasonality
trend = decomposition.trend

# Compute the gradient of the SSD time series
gradient = np.gradient(Tb_std['SSD'].values)

# Clean the gradient by removing NaN and infinite values
gradient_clean = gradient[~np.isnan(gradient) & ~np.isinf(gradient)]

# Check for empty or zero-variance gradient
if gradient_clean.size == 0:
    print("Error: Gradient array is empty!")
elif np.std(gradient_clean) == 0:
    print("Warning: No variance in the gradient. Thresholds may not be meaningful.")
else:
    # Compute the positive and negative thresholds
    pos_threshold = np.mean(gradient_clean) + 3 * np.std(gradient_clean)
    neg_threshold = np.mean(gradient_clean) - 3 * np.std(gradient_clean)

    print(f"Positive Threshold: {pos_threshold}")
    print(f"Negative Threshold: {neg_threshold}")

    # Find indices of values above the positive threshold
    pos_indices = np.where(gradient > pos_threshold)[0]

    # List to store the dates where positive is followed by negative threshold
    valid_anomaly_dates = []

    # Loop through positive anomalies and search for a negative value in the next 3 points
    #for pos_idx in pos_indices:
        #for offset in range(1, 7):  # Checking the next 3 points
            #next_idx = pos_idx + offset
            #if next_idx < len(gradient) and gradient[next_idx] < neg_threshold:
                #valid_anomaly_dates.append(Tb_std.index[pos_idx])
                #break  # Stop searching after finding the first valid negative gradient


    
    valid_anomaly_dates = Tb_std.index[pos_indices]
    # Remove duplicate dates
    valid_anomaly_dates = pd.to_datetime(valid_anomaly_dates).unique()
    anomaly_after_march = valid_anomaly_dates[valid_anomaly_dates > f"{year}-03-31"]

    # Print the detected valid anomaly dates
    print("Detected valid positive-to-negative threshold dates:")
    print(anomaly_after_march)

    # --- Plot Results ---
    fig, axes = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

    # Find the date and value of the maximum trend
    trend_after_march = trend[trend.index > f"{year}-03-31"]  # Adjust the year as needed or make it dynamic
    max_trend_value = trend_after_march.max()
    max_trend_date = trend_after_march.idxmax()

    # Calculate where the trend crosses the 60th percentile
    sign_change = (trend - percentile_60) * (trend.shift(-1) - percentile_60) < 0  # Strict sign change condition
    intersection_points = trend[sign_change]

    # --- Plot SSD on Primary Axis (Top Subplot) ---
    axes[0].plot(Tb_std.index, Tb_std['SSD'], color='#1b9e77', label='SSD Time Series', linewidth=2)
    axes[0].plot(Tb_std_Clean.index, trend, color='blue', label='SSD Trend Component', linewidth=2, linestyle='--')
    axes[0].axvline(x=max_trend_date, color='purple', linestyle='--', linewidth=3, 
                label=f'Max Trend on {max_trend_date.strftime("%Y-%m-%d")}', zorder=4)
    axes[0].axhline(y=percentile_60, color='navy', linestyle='--', 
                    label=f'CETB SSD 60th Percentile: {percentile_60:.2f} K', linewidth=2)
    axes[0].axhline(y=percentile_99, color='tab:pink', linestyle='--', 
                    label=f'CETB SSD 99.3th Percentile: {percentile_99:.2f} K', linewidth=2)
    # Highlight intersection points with circles
    filtered_intersection = intersection_points[(intersection_points.index > f"{year}-03-31") & (intersection_points.index < f"{year}-09-01")]
    axes[0].scatter(filtered_intersection.index, filtered_intersection.values, color='red', s=120, 
                    label='Trend & 60th Percentile Intersections', edgecolor='black', zorder=5)

    axes[0].axvline(x=MeltDate, color='red', linestyle='--', label='2003-04-13')
    axes[0].set_ylabel('SSD (K)', fontsize=14, color='#1b9e77')
    axes[0].grid(True)

    # Highlight valid anomaly points on the SSD time series plot
    intersection_date = filtered_intersection.index[0]  # Assuming we want the first intersection date

    # Find the maximum trend after March of the given year
    trend_after_march = trend[trend.index > f"{year}-03-31"]
    if trend_after_march.empty:
        print("No trend data available after March.")
    else:
        max_trend_value = trend_after_march.max()
        max_trend_date = trend_after_march.idxmax()

        # Convert the max trend date to its index in the gradient array
        if max_trend_date in Tb_std.index:
            max_trend_idx = Tb_std.index.get_loc(max_trend_date)

            # Get the SSD value on the day of the max trend
            max_trend_ssd = Tb_std.loc[max_trend_date, 'SSD']

            # Check if the gradient at or near this date is above the positive threshold
            window = 14  # Check up to the next 14 points/7 days for a sudden increase
            sudden_increase_found = False
#### Add codes below because it should be after the max trend location
            for offset in range(0, window + 1):
                check_idx = max_trend_idx + offset
                if check_idx < len(gradient) and gradient[check_idx] > pos_threshold:
                    increase_date = Tb_std.index[check_idx]
                    increase_ssd = Tb_std.loc[increase_date, 'SSD']

                    # Search the SSD values on the sudden gradient increase date, the day before, and the day after
                    max_ssd_date = increase_date
                    max_ssd_value = increase_ssd

                    # Check the day before the gradient spike (-1 day), if needed
                    prev_day_idx = check_idx - 1
                    if prev_day_idx >= 0:
                        prev_increase_date = Tb_std.index[prev_day_idx]
                        prev_increase_ssd = Tb_std.loc[prev_increase_date, 'SSD']
                        if prev_increase_ssd > max_ssd_value:
                            max_ssd_value = prev_increase_ssd
                            max_ssd_date = prev_increase_date

                    # Check the day after the gradient spike (+1 day)
                    next_day_idx = check_idx + 1
                    if next_day_idx < len(Tb_std):
                        next_increase_date = Tb_std.index[next_day_idx]
                        next_increase_ssd = Tb_std.loc[next_increase_date, 'SSD']
                        if next_increase_ssd > max_ssd_value:
                            max_ssd_value = next_increase_ssd
                            max_ssd_date = next_increase_date

                        # Print the results for max SSD within the 3-day window
                    print(f"Max trend after March occurs on {max_trend_date.strftime('%Y-%m-%d %H:%M:%S')} "
                          f"(Trend Value: {max_trend_value:.2f}, SSD: {max_trend_ssd:.2f}).")
                    print(f"Sudden gradient increase detected on {increase_date.strftime('%Y-%m-%d %H:%M:%S')} (SSD: {increase_ssd:.2f}).")
                    print(f"Actual SSD increase occurs on {max_ssd_date.strftime('%Y-%m-%d %H:%M:%S')} "
                          f"(Max SSD: {max_ssd_value:.2f}).")
                
                    sudden_increase_found = True
                    break

            if not sudden_increase_found:
                print(f"Max trend after March occurs on {max_trend_date.strftime('%Y-%m-%d')} "
                      f"(Trend Value: {max_trend_value:.2f}, SSD: {max_trend_ssd:.2f}), but no sudden gradient increase found within {window} steps.")
        else:
            print("Max trend date not found in the dataset index.")

##### The anomaly should be the max of sharp increase in gradients or the day after or before it
    axes[0].scatter(max_ssd_date, Tb_std.loc[max_ssd_date, 'SSD'], color='tab:pink', marker="D", s=100, label=f'First Sharp SSD Increase after Max Trend: {max_ssd_value:.2f} K')
    axes[0].legend(loc='center right')

    # Add Tb and DAV on Secondary Axis (Right)
    ax2 = axes[0].twinx()
    ax2.plot(DAV_df.index, DAV_df['DAV'], label='DAV Time Series', color='black', linewidth=2)
    ax2.plot(Tb_nearest.index, Tb_nearest['TB'], label='TB Time Series', color='orange', linewidth=1.5)
    ax2.plot(swe_df.index, swe_df['value'], label='SWE (mm)', color='purple', linewidth=3, zorder=1)
    ax2.axvline(x=max_swe_date, color='black', linestyle='--', label=f'Max SWE on day {max_swe_day_of_year}', linewidth=2)
    ax2.plot(filtered_temp_df.index, filtered_temp_df['value'], label='Temp (K)', color='tab:red', zorder=2)
    ax2.axhline(y=268.15, color='yellow', linestyle='--', linewidth=2, label='Air Temperature Threshold (268.15 K)')
    ax2.axhline(y=Tb_threshold, color='gray', linestyle='--', 
            label=f'TB Threshold: {Tb_threshold} K', linewidth=2)
    ax2.axhline(y=DAV_threshold, color='red', linestyle='--', 
            label=f'DAV Threshold (90th): {DAV_threshold:.2f} K', linewidth=2)
    ax2.set_ylabel('DAV/TB (K)', fontsize=14, color='black')
    ax2.legend(loc='upper right')

    # --- Plot Gradient on Bottom Subplot ---
    # Use real dates for the gradient x-axis
    gradient_dates = Tb_std.index
    axes[1].plot(gradient_dates, gradient, color='tab:blue', label='Gradient', linewidth=1.5)
    axes[1].axhline(y=pos_threshold, color='green', linestyle='--', 
                    label=f'Positive Threshold: {pos_threshold:.2f}')
    #axes[1].axhline(y=neg_threshold, color='red', linestyle='--', 
                    #label=f'Negative Threshold: {neg_threshold:.2f}')
    axes[1].set_title('Gradient of SSD Time Series', fontsize=16)
    axes[1].set_ylabel('Gradient', fontsize=14)
    axes[1].grid(True)

    # Highlight valid anomaly points on the gradient plot
    validanamoly = anomaly_after_march[anomaly_after_march < intersection_date]
    valid_indices = [Tb_std.index.get_loc(date) for date in validanamoly]
    valid_indices = valid_indices[0]
    axes[1].scatter(gradient_dates[valid_indices], gradient[valid_indices], 
                    color='tab:pink', marker="D", s=100, label='First Sharp Increase Exceeding Positive Threshold of Gradient, after Max SSD Trend')
    axes[1].legend()

    # Format x-axis for date alignment
    axes[1].set_xlabel('Date', fontsize=14)
    
    plt.title(f'SSD Analysis for Snowmelt Detection, {Site}, {year}')  # Plot title

    # Adjust layout and show the plots
    plt.tight_layout()
    
    # Save the plot to the specified path (include bbox_extra_artists)
    plt.savefig(save_path, bbox_inches='tight')
    plt.show()
```
